```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 

df_sinasc = pd.read_csv('Profissão Cientista de Dados M6 - SINASC_RO_2019.csv')
```

####  1) Na base do SINASC do DataSUS, faça um boxplot da idade da mãe.

```python
df_sinasc.boxplot(column='IDADEMAE', figsize=(6,8))

plt.title('Distribuição da Idade da Mãe')
plt.ylabel('Idade (Anos)')
plt.show()
```

####  2) Faça um boxplot do peso do bebê

```python
plt.figure(figsize=(6,8))
df_sinasc.boxplot(column='PESO')

plt.title('Distribuição do Peso dos Bebês (SINASC RO 2019)')
plt.ylabel('Peso em gramas')
plt.show()
```

####  3) Faça um histograma da idade da mãe

```python
plt.figure(figsize=(8, 6))
df_sinasc['IDADEMAE'].plot.hist(bins=20, color='pink', edgecolor='white')

plt.title('Histograma: Idade da Mãe (SINASC RO 2019)')
plt.xlabel('Idade (Anos)')
plt.ylabel('Frequência (Quantidade de Nascimentos)')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

#### 4) Faça um histograma do peso do bebê

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_sinasc['PESO'], bins=30, kde=True, color='darkblue')

plt.title('Distribuição do Peso dos Bebês (SINASC RO 2019)', fontsize=14)
plt.xlabel('Peso (gramas)', fontsize=12)
plt.ylabel('Frequência de Nascimentos', fontsize=12)

plt.grid(axis='y', alpha=0.3)

plt.show()
```
