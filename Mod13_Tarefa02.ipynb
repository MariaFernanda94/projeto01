```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score

df = pd.read_csv('previsao_de_renda.csv')
df.drop(['Unnamed: 0', 'data_ref', 'id_cliente'], axis=1, inplace=True)
df.dropna(inplace=True) # Removendo nulos para simplificar o fluxo dos modelos lineares

df_dummies = pd.get_dummies(df, drop_first=True)
X = df_dummies.drop('renda', axis=1)
y = df_dummies['renda']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
alphas = [0, 0.001, 0.005, 0.01, 0.05, 0.1]

print("--- RESULTADOS RIDGE ---")
for a in alphas:
    modelo_ridge = Ridge(alpha=a)
    modelo_ridge.fit(X_train, y_train)
    y_pred = modelo_ridge.predict(X_test)
    print(f"Alpha {a}: R² = {r2_score(y_test, y_pred):.4f}")

print("\n--- RESULTADOS LASSO ---")
for a in alphas:
    modelo_lasso = Lasso(alpha=a)
    modelo_lasso.fit(X_train, y_train)
    y_pred = modelo_lasso.predict(X_test)
    print(f"Alpha {a}: R² = {r2_score(y_test, y_pred):.4f}")
```

```python
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import LinearRegression

reg_linear = LinearRegression()
sfs = SequentialFeatureSelector(reg_linear, n_features_to_select='auto', direction='forward', cv=5)
sfs.fit(X_train, y_train)

X_train_step = sfs.transform(X_train)
X_test_step = sfs.transform(X_test)

reg_linear.fit(X_train_step, y_train)
y_pred_step = reg_linear.predict(X_test_step)
print(f"R² Stepwise (Forward): {r2_score(y_test, y_pred_step):.4f}")
```

```python
df_coef = pd.DataFrame({
    'Variável': X.columns,
    'Ridge (alpha=0.1)': modelo_ridge.coef_,
    'Lasso (alpha=0.1)': modelo_lasso.coef_
})

df_coef['dif_absoluta'] = (df_coef['Ridge (alpha=0.1)'] - df_coef['Lasso (alpha=0.1)']).abs()

print("--- COMPARAÇÃO DE PARÂMETROS (COEFICIENTES) ---")
print(df_coef.sort_values(by='dif_absoluta', ascending=False))

print(f"\nScore R² Ridge: {r2_score(y_test, modelo_ridge.predict(X_test)):.4f}")
print(f"Score R² Lasso: {r2_score(y_test, modelo_lasso.predict(X_test)):.4f}")
```

```python
y_train_log = np.log(y_train)
y_test_log = np.log(y_test)

modelo_melhorado = Ridge(alpha=0.01)
modelo_melhorado.fit(X_train, y_train_log)

y_pred_log = modelo_melhorado.predict(X_test)
y_pred_original = np.exp(y_pred_log)

print(f"R² após transformação logarítmica: {r2_score(y_test, y_pred_original):.4f}")
```

```python
arvore = DecisionTreeRegressor(max_depth=5, random_state=42)
arvore.fit(X_train, y_train)

y_pred_tree = arvore.predict(X_test)
print(f"R² Árvore de Regressão: {r2_score(y_test, y_pred_tree):.4f}")
```
