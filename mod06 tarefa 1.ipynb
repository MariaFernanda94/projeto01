#### 1) A base 'sinasc_RO_2019.csv' contém dados do sistema de nascidos vivos do DataSUS, cujo dicionário de dados está no arquivo 'Estrutura_SINASC_para_CD.pdf'. Carregue esses arquivos.

```python
import pandas as pd

# Carregando a base da dados
df_sinasc = pd.read_csv('Profissão Cientista de Dados M6 - SINASC_RO_2019.csv')

#Visualizando as primeiras 5 linhas para conferência 
print("Dimensôes da base:", df_sinasc.shape)
df_sinasc.head()
```

####  2) Monte um gráficos de barras com as variáveis APGAR1, APGAR5, SEXO, GESTACAO, ESTCIVMAE.

```python
import matplotlib.pyplot as plt 
import seaborn as sns

```

```python
# Lista das colunas que você quer
colunas = ['APGAR1', 'APGAR5', 'SEXO', 'GESTACAO', 'ESTCIVMAE']

for col in colunas:
    plt.figure(figsize=(5, 3))
    df_sinasc[col].value_counts().sort_index().plot.bar()
    
    plt.title(f'Frequência de {col}')
    plt.ylabel('Quantidade de Nascidos')
    plt.xticks(rotation=45)
    plt.show()
```

#### 3) Monte um gráfico de barras mostrando o APGAR5 médio por CONSULTAS

```python
apgar_medio = df_sinasc.groupby('CONSULTAS')['APGAR5'].mean()

plt.figure(figsize=(8, 5))
apgar_medio.plot.bar(color='red')

plt.title('Média do APGAR5 por Quantidade de Consultas Pré-natais')
plt.xlabel('Código de Consultas')
plt.ylabel('Média do APGAR5')
plt.xticks(rotation=0) # Deixa os números do eixo X em pé
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.ylim(8, 10) 

plt.show()
```

####  4) Monte um painel com 4 gráficos mostrando as frequencias de SEXO, GESTACAO, ESTCIVMAE e CONSULTAS

```python
variaveis = ['SEXO', 'GESTACAO', 'ESTCIVMAE', 'CONSULTAS']


fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for i, col in enumerate(variaveis):
    # Calculamos em qual linha e coluna o gráfico deve entrar
    linha = i // 2
    coluna = i % 2
    
    df_sinasc[col].value_counts().sort_index().plot.bar(ax=axes[linha, coluna], color='teal')
    
    axes[linha, coluna].set_title(f'Frequência de {col}', fontweight='bold')
    axes[linha, coluna].set_ylabel('Quantidade')
    axes[linha, coluna].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
