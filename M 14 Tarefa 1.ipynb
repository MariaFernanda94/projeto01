```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

sns.set_theme(style="whitegrid")
```

```python
meses = ['MAR', 'ABR', 'MAI', 'JUN', 'DEZ']

for mes in meses:
    # Ajuste o nome do arquivo conforme o padrão da imagem 2
    nome_arquivo = f'SINASC_RO_2019_{mes}.csv'
    
    try:
        # Lendo o arquivo do mês
        sinasc = pd.read_csv(nome_arquivo)
        
        # Criando o gráfico (Exemplo: Escolaridade da mãe vs Idade da mãe)
        # Usando a estrutura das aulas (figsize e plotagem)
        plt.figure(figsize=(15, 5))
        sns.boxplot(data=sinasc, x='ESCMAE', y='IDADEMAE')
        
        plt.title(f'Idade da Mãe por Escolaridade - Mês: {mes}')
        plt.ylabel('Idade da Mãe')
        plt.xlabel('Escolaridade')
        
        # Seguindo a ideia de organização das aulas: criar uma pasta para os outputs
        os.makedirs('output', exist_ok=True)
        plt.savefig(f'./output/grafico_{mes}.png')
        plt.show()
        
    except FileNotFoundError:
        print(f"Alerta: O arquivo {nome_arquivo} não foi localizado na pasta atual.")
```
